## Exercise 2: CSV, dataframe, plot

```python
csv_text = """size,cpu_s,gpu_s
10,0.00001,0.00006
100,0.00188,0.00338
1000,0.00902,0.00128
4000,0.41043,0.09528"""

with open("benchmark.csv", "w") as f:
    f.write(csv_text)
```

```python
import pandas as pd
df = pd.read_csv("benchmark.csv")
df
```

```python
import matplotlib.pyplot as plt
plt.plot(df["size"], df["cpu_s"], label="CPU")
plt.plot(df["size"], df["gpu_s"], label="GPU")
plt.xlabel("matrix size")
plt.ylabel("seconds")
plt.legend()
plt.xscale("log")
plt.yscale("log")
plt.show()
```
